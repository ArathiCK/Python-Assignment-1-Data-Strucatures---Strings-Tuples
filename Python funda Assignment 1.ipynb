## --- Question 1: String Concatenation ---

```python
string1 = "Hello"
string2 = input("Enter your Name: ")
output_1a = string1 +" "+string2
print(output_1a)
```

```python
string3 = ", welcome to python programming"
output_1b = output_1a + string3
print(output_1b)
```

## --- Question 2: String Slicing and Indexing ---

```python
# a. Print first character 
print(output_1b[0])
# b. Print last character
print(output_1b[-1])
# c. Print first 5 characters
print(output_1b[0:5])
# d. Print last 11 characters
print(output_1b[-11:])
# e. Print string in reverse
print(output_1b[::-1])
# f. Print "Python"
print(output_1b[-18:-11])
```

## --- Tuples Section ---

```python
tuple1 = (10, 20, 30)
tuple2 = (40, 50, 60)
# a. Concatenate the two tuples
t_combine = tuple1 + tuple2
print(t_combine)
```

```python
# b. Repeat elements of "t_combine" 3 times
t_repeat = t_combine * 3
print(t_repeat)
```

```python
# c. Access 3rd element
print(t_combine[2])
```

```python
# d. Access first three elements
print(t_combine[0:3])
```

```python
# e. Access last three elements
print(t_combine[-3:])
```
